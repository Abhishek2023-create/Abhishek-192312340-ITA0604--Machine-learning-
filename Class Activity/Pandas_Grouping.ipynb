```python
import pandas as pd

# Create data
data = {
    "Department": ["CSE", "CSE", "ECE", "ECE"],
    "Marks": [80, 70, 90, 60]
}

# Create DataFrame
df = pd.DataFrame(data)

# Average marks department-wise
print("Average Marks:")
print(df.groupby("Department")["Marks"].mean())

# Total marks department-wise
print("\nTotal Marks:")
print(df.groupby("Department")["Marks"].sum())

# Maximum marks department-wise
print("\nMaximum Marks:")
print(df.groupby("Department")["Marks"].max())

# Minimum marks department-wise
print("\nMinimum Marks:")
print(df.groupby("Department")["Marks"].min())

# Number of students department-wise
print("\nStudent Count:")
print(df.groupby("Department")["Marks"].count())
```
