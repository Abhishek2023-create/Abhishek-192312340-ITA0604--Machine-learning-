```python
# ==========================================
# SIMPLE MISTAKE BOUND MODEL
# Loan Approval Example
# ==========================================

# Applicant income in thousands
X = [10, 15, 20, 30, 40, 50]

# Actual loan decision
# -1 = REJECTED
#  1 = APPROVED

y = [-1, -1, -1, 1, 1, 1]

# Initially model knows nothing
weight = 0

# Count mistakes
mistakes = 0

print("REJECTED = -1")
print("APPROVED = +1")
print("--------------------------")

# Check each applicant
for i in range(len(X)):

    # Model prediction
    score = weight * X[i]

    if score >= 0:
        prediction = 1       # APPROVED
    else:
        prediction = -1      # REJECTED

    # Convert numbers to names
    actual_name = "APPROVED" if y[i] == 1 else "REJECTED"

    predicted_name = "APPROVED" if prediction == 1 else "REJECTED"

    print("\nApplicant", i + 1)

    print("Income :", X[i], "thousand")

    print("Actual :", actual_name)

    print("Prediction :", predicted_name)

    # Check mistake
    if prediction != y[i]:

        mistakes += 1

        print("MISTAKE")

        # Learn from mistake
        weight = weight + y[i] * X[i]

        print("New Weight =", weight)

    else:

        print("CORRECT")

# Display total mistakes
print("\n========================")

print("TOTAL MISTAKES =", mistakes)

print("========================")
```
