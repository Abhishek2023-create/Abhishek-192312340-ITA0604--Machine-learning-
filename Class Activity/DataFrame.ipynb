```python
import pandas as pd

# Create data
data = {
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Marks": [80, 65, 90, 75],
    "City": ["Chennai", "Madurai", "Salem", "Chennai"]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display complete DataFrame
print("Student Data:")
print(df)

# Access one column
print("\nNames:")
print(df["Name"])

# Access multiple columns
print("\nNames and Marks:")
print(df[["Name", "Marks"]])
```
