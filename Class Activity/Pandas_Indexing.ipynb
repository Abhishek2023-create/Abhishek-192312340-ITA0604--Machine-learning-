```python
import pandas as pd

# Create DataFrame
data = {
    "Name": ["Arun", "Bala", "Cathy"],
    "Marks": [80, 65, 90],
    "City": ["Chennai", "Madurai", "Salem"]
}

df = pd.DataFrame(data)

# Display DataFrame
print("Student Data:")
print(df)

# Using loc - select row by label
print("\nFirst Row using loc:")
print(df.loc[0])

# Using loc - select specific row and column
print("\nMarks of First Student:")
print(df.loc[0, "Marks"])

# Using iloc - select row by position
print("\nFirst Row using iloc:")
print(df.iloc[0])

# Using iloc - select first row, second column
print("\nFirst Row, Second Column:")
print(df.iloc[0, 1])
```
