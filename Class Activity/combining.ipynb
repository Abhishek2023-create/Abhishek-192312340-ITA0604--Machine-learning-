```python
import pandas as pd

# Create sales data
data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Tablet"],
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Electronics", "Electronics", "Electronics"],
    "Sales": [50000, 30000, 20000, 45000, 35000, 25000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Group by Category and calculate total and average
result = df.groupby("Category")["Sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean"
)

# Display result
print("Sales Summary:")
print(result)
```
