```python
import pandas as pd

# Create student data
data = {
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Marks": [80, 65, 90, 75],
    "City": ["Chennai", "Madurai", "Salem", "Chennai"]
}

df = pd.DataFrame(data)

# Marks greater than 70
print("Marks > 70:")
print(df[df["Marks"] > 70])

# Marks greater than or equal to 75
print("\nMarks >= 75:")
print(df[df["Marks"] >= 75])

# Students from Chennai
print("\nStudents from Chennai:")
print(df[df["City"] == "Chennai"])

# Marks > 70 AND Chennai
print("\nMarks > 70 AND Chennai:")
print(df[(df["Marks"] > 70) & (df["City"] == "Chennai")])

# Marks > 80 OR Chennai
print("\nMarks > 80 OR Chennai:")
print(df[(df["Marks"] > 80) | (df["City"] == "Chennai")])
```
