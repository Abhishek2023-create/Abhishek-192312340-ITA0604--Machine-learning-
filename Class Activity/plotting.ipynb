```python
import pandas as pd
import matplotlib.pyplot as plt

# Create DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Marks": [80, 65, 90, 75]
})

# Bar Chart
df.plot(x="Name", y="Marks", kind="bar")
plt.title("Student Marks - Bar Chart")
plt.show()

# Line Chart
df.plot(x="Name", y="Marks", kind="line")
plt.title("Student Marks - Line Chart")
plt.show()

# Pie Chart
df.plot(
    y="Marks",
    kind="pie",
    labels=df["Name"],
    autopct="%1.1f%%"
)
plt.title("Student Marks - Pie Chart")
plt.show()

# Histogram
df["Marks"].plot(kind="hist")
plt.title("Distribution of Marks")
plt.show()
```
